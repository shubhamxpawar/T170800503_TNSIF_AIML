# Day 2: Control Structures and Functions

## If-Else Statements

```
num = 7
if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

## Elif

```
marks = 75
if marks >= 90:
    print("A")
elif marks >= 75:
    print("B")
elif marks >= 60:
    print("C")
else:
    print("F")
```

## For Loop

```
for i in range(1, 6):
    print(i)
```

## While Loop

```
count = 1
while count <= 5:
    print(count)
    count += 1
```

## Break and Continue

```
for i in range(1, 6):
    if i == 3: break
    print(i)
```

## Functions

```
def add(a,b):
    return a+b

print(add(5,3))
```

## String Slicing

```
s = "Python"
print(s[0:3], s[::-1])
```

## Simple Programs

```
def fact(n):
    res = 1
    for i in range(1,n+1):
        res *= i
    return res
print(fact(5))
```
