```python
num = int(input("Enter a number: "))
reversed_num = 0
temp = abs(num)
while temp > 0:
    digit = temp % 10
    reversed_num = (reversed_num * 10) + digit
    temp //= 10
if num < 0:
    reversed_num = -reversed_num
print("Reversed number:", reversed_num)
```

```python
num = int(input("Enter a number: "))
is_prime = True
if num <= 1:
    is_prime = False
else:
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
if is_prime:
    print(num, "is a prime number")
else:
    print(num, "is not a prime number")
```

```python
def is_palindrome(num):
    return str(num) == str(num)[::-1]

val = int(input("Enter a number: "))
print(is_palindrome(val))
```

```python
def sum_of_digits(num):
    return sum(int(digit) for digit in str(abs(num)))

val = int(input("Enter a number: "))
print("Sum of digits:", sum_of_digits(val))
```

### 🔹 List, Tuple, Set (11–20)

```python
numbers = [10, 20, 30, 40, 50]
total = sum(numbers)
average = total / len(numbers)
print("Sum:", total)
print("Average:", average)
```

```python
numbers = [45, 12, 89, 5, 23]
smallest = numbers[0]
largest = numbers[0]
for num in numbers:
    if num < smallest:
        smallest = num
    if num > largest:
        largest = num
print("Smallest:", smallest)
print("Largest:", largest)
```

```python
numbers = [1, 2, 2, 3, 4, 4, 5]
unique_list = []
for num in numbers:
    if num not in unique_list:
        unique_list.append(num)
print(unique_list)
```

```python
numbers = [1, 2, 3, 4, 5]
reversed_list = []
for i in range(len(numbers) - 1, -1, -1):
    reversed_list.append(numbers[i])
print(reversed_list)
```

### 🔹 Dictionary (21–25)

```python
students = {
    "Alice": 85,
    "Bob": 92,
    "Charlie": 78,
    "David": 95
}

best_student = max(students, key=students.get)
print("Highest marks:", best_student, "-", students[best_student])
```

```python
text = "hello world"
frequency = {}
for char in text:
    frequency[char] = frequency.get(char, 0) + 1
print(frequency)
```

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 3, "c": 4}
merged_dict = {**dict1, **dict2}
print(merged_dict)
```

```python
my_dict = {"apple": 5, "banana": 2, "cherry": 8, "date": 1}
sorted_dict = dict(sorted(my_dict.items(), key=lambda item: item[1]))
print(sorted_dict)
```

### 🔹 Classes & Objects (26–30)

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_details(self):
        print(f"Name: {self.name}, Age: {self.age}")

s1 = Student("John", 20)
s1.display_details()
```

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

rect = Rectangle(5, 10)
print("Area:", rect.area())
print("Perimeter:", rect.perimeter())
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def increase_salary(self, percentage):
        self.salary += self.salary * (percentage / 100)

emp = Employee("Sarah", 50000)
emp.increase_salary(10)
print("New Salary:", emp.salary)
```

```python
class ShoppingCart:
    def __init__(self):
        self.items = {}

    def add_item(self, item, price):
        self.items[item] = price

    def calculate_total(self):
        return sum(self.items.values())

cart = ShoppingCart()
cart.add_item("Book", 15)
cart.add_item("Pen", 2)
print("Total Bill:", cart.calculate_total())
```

```python
class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def remove_book(self, book):
        if book in self.books:
            self.books.remove(book)

lib = Library()
lib.add_book("1984")
lib.add_book("To Kill a Mockingbird")
lib.remove_book("1984")
print(lib.books)
```
