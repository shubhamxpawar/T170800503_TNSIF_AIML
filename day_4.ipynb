### 1. Import Libraries
Importing the required libraries for our tasks.

```python
import numpy as np
import pandas as pd
```

### 2. Create a NumPy Array of Marks
Creating a basic 1D array to store student marks.

```python
marks = np.array([65, 78, 90, 55, 88])
print(marks)
```

### 3. Calculate Array Statistics
Finding basic statistics like average, highest, and lowest marks.

```python
avg_marks = np.mean(marks)
highest_marks = np.max(marks)
lowest_marks = np.min(marks)

print("Average:", avg_marks)
print("Highest:", highest_marks)
print("Lowest:", lowest_marks)
```

### 4. Special Arrays
Generating arrays using arange, zeros, and ones.

```python
range_arr = np.arange(1, 10, 2)
zeros_arr = np.zeros((2, 3))
ones_arr = np.ones((3, 2))

print("Arange Array:", range_arr)
print("Zeros Array:\n", zeros_arr)
print("Ones Array:\n", ones_arr)
```

### 5. Identity Matrix
Creating a 3x3 identity matrix.

```python
identity_matrix = np.eye(3)
print(identity_matrix)
```

### 6. Reshape and Transpose
Changing the shape of an array and transposing it.

```python
original_arr = np.arange(1, 7)
reshaped_matrix = original_arr.reshape(2, 3)
transposed_matrix = reshaped_matrix.T

print("Reshaped:\n", reshaped_matrix)
print("Transpose:\n", transposed_matrix)
```

### 7. NumPy Array Operations
Performing basic arithmetic operations and sorting on an array.

```python
base_arr = np.array([30, 10, 20, 50, 40])
add_five = base_arr + 5
multiply_two = base_arr * 2
sorted_arr = np.sort(base_arr)

print("Original:", base_arr)
print("Added 5:", add_five)
print("Multiplied by 2:", multiply_two)
print("Sorted:", sorted_arr)
```

### 8. Create a Pandas Series
Converting our student marks into a labeled Series.

```python
student_labels = ['Student A', 'Student B', 'Student C', 'Student D', 'Student E']
marks_series = pd.Series(marks, index=student_labels)
print(marks_series)
```

### 9. Create a Pandas DataFrame
Building a DataFrame to store student records.

```python
student_data = {
    'Name': ['Amit', 'Bina', 'Chirag', 'Divya', 'Esha'],
    'Age': [20, 21, 19, 22, 20],
    'Marks': [65, 78, 90, 55, 88]
}
student_df = pd.DataFrame(student_data)
print(student_df)
```

### 10. Display and Summary
Viewing the first few rows and structure of the DataFrame.

```python
print("First 5 Rows:")
print(student_df.head())
print("\nDataFrame Info:")
student_df.info()
```

### 11. Selection using Loc and Iloc
Accessing rows and columns by labels and positions.

```python
names_col = student_df.loc[:, 'Name']
first_row = student_df.iloc[0]

print("Names Column:\n", names_col)
print("\nFirst Row:\n", first_row)
```

### 12. Filtering Data
Filtering students who scored above 75 marks.

```python
high_scorers = student_df[student_df['Marks'] > 75]
print(high_scorers)
```

### 13. Handling Missing Values
Handling null values using dropna and fillna.

```python
null_df = pd.DataFrame({
    'A': [1.0, 2.0, None, 4.0],
    'B': [5.0, None, None, 8.0]
})

print("Original with NaNs:\n", null_df)
print("\nDropped NaNs:\n", null_df.dropna())
print("\nFilled NaNs with 0.0:\n", null_df.fillna(0.0))
```

### 14. Sorting and Groupby
Sorting our student DataFrame by Marks and calculating average marks per group.

```python
sorted_students = student_df.sort_values(by='Marks')
print("Sorted by Marks:\n", sorted_students)

student_df['Category'] = ['Batch A', 'Batch B', 'Batch A', 'Batch B', 'Batch A']
group_avg = student_df.groupby('Category')['Marks'].mean()
print("\nAverage Marks by Category:\n", group_avg)
```
