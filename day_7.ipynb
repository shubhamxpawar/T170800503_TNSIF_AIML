# Matplotlib Basics

## What is Matplotlib
Matplotlib is a Python library used for data visualization. It converts numbers and data into charts so that trends and patterns become easy to understand at a glance. It works hand in hand with NumPy and Pandas and gives full control over figure size, labels, legends, colours and styles.

## Where Matplotlib is Used
- Exploratory Data Analysis of any dataset
- Studying trends and patterns in data
- Dashboards and reports
- Research work in science, finance and engineering
- Machine learning result visualization

## Common Matplotlib Functions

| Function | Purpose |
|---|---|
| `plt.plot(x, y)` | Line plot |
| `plt.bar(x, y)` | Bar chart |
| `plt.hist(data, bins)` | Histogram |
| `plt.scatter(x, y)` | Scatter plot |
| `plt.pie(values, labels)` | Pie chart |
| `plt.xlabel()` / `plt.ylabel()` | Axis labels |
| `plt.title()` | Plot title |
| `plt.legend()` | Legend box |
| `plt.grid()` | Grid lines |
| `plt.subplot()` | Several plots in one figure |
| `plt.figure(figsize=(w, h))` | Set figure size |
| `plt.show()` | Display the plot |

## First Line Plot
Calling NumPy for sample data and drawing a dashed line plot with labels, a title, a legend and a grid.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 10)
y = np.sin(x)

print("x values:", x)
print("y values:", y)

plt.plot(x, y, label="sin(x)", color="blue", linestyle="--")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Sine Wave")
plt.legend()
plt.grid(True)
plt.show()
```

## Line Plot with Two Curves
Two `plot()` calls on the same axes draw sin(x) and cos(x) together in different colours.

```python
y2 = np.cos(x)

print("cos(x):", y2)

plt.plot(x, y, label="sin(x)", color="red")
plt.plot(x, y2, label="cos(x)", color="green")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Sine and Cosine Waves")
plt.legend()
plt.show()
```

## Bar Chart
A bar chart compares values across categories. Each bar gets its own colour.

```python
categories = ['A', 'B', 'C', 'D']
values = [10, 24, 36, 18]

plt.bar(categories, values, color=['blue', 'orange', 'green', 'red'], label="categories")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Bar Chart")
plt.legend()
plt.show()
```

## Histogram of Random Data
A histogram groups numbers into bins and shows how many values fall in each bin. Here 100 random values from a normal distribution are used.

```python
data = np.random.randn(100)

print("Random data:", data[:10], "...")

plt.hist(data, bins=10, color='purple', edgecolor='black', label="random")
plt.title("Histogram of Random Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

## Histogram with Custom Bins (Student Marks)
Bins can be given as explicit ranges instead of a count, which is useful for grouped marks.

```python
marks = [10, 15, 25, 30, 35, 40, 45, 50, 52, 55, 60, 42, 48, 33, 28, 22, 18]
bins = [0, 20, 40, 60]

plt.hist(marks, bins=bins, edgecolor='black')
plt.xlabel("Marks Range")
plt.ylabel("Number of Students")
plt.title("Histogram of Student Marks")
plt.show()
```

## Scatter Plot
A scatter plot shows individual points, which helps to see relationships between two sets of values.

```python
categories = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I']
values = [10, 24, 36, 18, 19, 20, 23, 26, 12]

plt.scatter(categories, values, color='darkgreen', label="categories")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Scatter Plot")
plt.legend()
plt.show()
```

## Pie Chart
A pie chart shows parts of a whole. `autopct` prints the percentage on each slice and `startangle` rotates the chart.

```python
sizes = [60, 20, 25, 25, 50, 45, 23]
labels = ['Apple', 'Banana', 'Mango', 'Grapes', 'A', 'B', 'C']

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90)
plt.title("Pie Chart")
plt.show()
```

## Subplots: Four Charts in One Figure
`plt.subplots()` creates a grid of axes. Each chart is drawn on its own axis with `set_title()` giving it a heading.

```python
fig, axs = plt.subplots(2, 2, figsize=(8, 8))

axs[0, 0].plot(x, y, color='blue')
axs[0, 0].set_title("Line Plot")

axs[0, 1].bar(categories, values, color='orange')
axs[0, 1].set_title("Bar Chart")

axs[1, 0].hist(marks, bins=bins, color='green')
axs[1, 0].set_title("Histogram")

axs[1, 1].scatter(categories, values, color='red')
axs[1, 1].set_title("Scatter Plot")

plt.tight_layout()
plt.show()
```

# Seaborn

## What is Seaborn
Seaborn is a visualization library built on top of Matplotlib. It offers a higher-level interface, so attractive and informative statistical charts need far fewer lines of code.

## Where Seaborn is Used
- Good-looking statistical charts with minimal code
- Fast data exploration and analysis
- Direct plotting from Pandas DataFrames
- Advanced colour palettes and built-in themes
- Complex charts like heatmaps, categorical plots and regression plots in one line

## Common Seaborn Functions

| Function | Purpose |
|---|---|
| `sns.scatterplot()` | Scatter plot |
| `sns.lineplot()` | Line plot |
| `sns.barplot()` | Averages with confidence intervals |
| `sns.countplot()` | Frequency of categories |
| `sns.boxplot()` | Box and whisker plot |
| `sns.violinplot()` | Boxplot combined with KDE |
| `sns.heatmap()` | Correlation matrix or tabular data |
| `sns.pairplot()` | Pairwise relationships in a dataset |
| `sns.histplot()` | Distribution of a variable |
| `sns.regplot()` / `sns.lmplot()` | Scatter plot with regression line |
| `sns.catplot()` | Categorical plots |
| `sns.jointplot()` | Bivariate plot with marginals |

## Loading the Tips Dataset
Seaborn ships with small built-in datasets. The tips dataset has restaurant bills with total bill, tip, day, time, gender and party size.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

tips = sns.load_dataset("tips")
print(tips.head())
```

## Scatter Plot: Total Bill vs Tip
`hue` colours the points by one column and `style` changes the marker shape by another.

```python
plt.figure(figsize=(6, 4))
sns.scatterplot(x="total_bill", y="tip", data=tips, hue="sex", style="time")
plt.title("Total Bill vs Tip")
plt.show()
```

## Line Plot: Total Bill vs Party Size
A line plot connects aggregated values and shows how the average bill changes with group size.

```python
plt.figure(figsize=(6, 4))
sns.lineplot(x="size", y="total_bill", data=tips, ci=None)
plt.title("Total Bill by Party Size")
plt.show()
```

## Bar Plot: Average Bill per Day
A seaborn bar plot shows the mean of the y column for each x category, with `hue` splitting each bar.

```python
plt.figure(figsize=(6, 4))
sns.barplot(x="day", y="total_bill", data=tips, hue="sex")
plt.title("Average Total Bill per Day")
plt.show()
```

## Count Plot: Customers per Day
A count plot counts how many rows belong to each category.

```python
plt.figure(figsize=(6, 4))
sns.countplot(x="day", data=tips, hue="sex")
plt.title("Number of Customers per Day")
plt.show()
```

## Box Plot: Bill Distribution per Day
The box shows the median and quartiles while the whiskers show the rest of the spread, including outliers.

```python
plt.figure(figsize=(6, 4))
sns.boxplot(x="day", y="total_bill", data=tips, hue="sex")
plt.title("Total Bill Distribution per Day")
plt.show()
```

## Violin Plot: Distribution by Day and Gender
A violin plot shows the density of the data. With `split=True` both hue categories share the same violin for easy comparison.

```python
plt.figure(figsize=(6, 4))
sns.violinplot(x="day", y="total_bill", data=tips, hue="sex", split=True)
plt.title("Bill Distribution by Day and Gender")
plt.show()
```

## Correlation Heatmap
A heatmap shows a matrix of numbers as colours. The correlation matrix reveals how strongly numeric columns move together.

```python
plt.figure(figsize=(6, 4))
corr = tips.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Heatmap of Tips Data")
plt.show()
```

## Pairplot
A pairplot draws a scatter plot for every pair of numeric columns and a distribution on the diagonal, coloured by a category with `hue`.

```python
sns.pairplot(tips, hue="sex")
plt.show()
```

## Distribution Plot
`histplot` with `kde=True` overlays a smooth density curve on the histogram.

```python
plt.figure(figsize=(6, 4))
sns.histplot(tips["total_bill"], kde=True, bins=20)
plt.title("Distribution of Total Bill")
plt.show()
```

## Regression Plot
`regplot` fits and draws a straight regression line over the scatter of two columns.

```python
plt.figure(figsize=(6, 4))
sns.regplot(x="total_bill", y="tip", data=tips)
plt.title("Regression: Tip vs Total Bill")
plt.show()
```

## LM Plot with Regression Line Split by Gender
`lmplot` is the figure-level version of regplot. With `hue` it draws one regression line per gender.

```python
sns.lmplot(x="total_bill", y="tip", data=tips, hue="sex", aspect=1.5)
plt.title("Total Bill vs Tip with Regression Line")
plt.show()
```

## Average Bill per Day Split by Gender

```python
sns.barplot(x="day", y="total_bill", hue="sex", data=tips)
plt.title("Average Total Bill per Day (Male vs Female)")
plt.show()
```
